```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/kaggle/input/datasets/aleespinosa/apple-watch-and-fitbit-data/aw_fb_data.csv')
print(df.shape)
df.head()
```

```python
# Basic info
df.info()
```

```python
# Null check
df.isnull().sum()
```

```python
# Basic stats
df.describe()
```

```python
# Value counts
print(df['activity'].value_counts())
print(df['device'].value_counts())
print(df['gender'].value_counts())
```

```python
# KPI values
avg_heart_rate = round(df['hear_rate'].mean(), 2)
avg_steps      = round(df['steps'].mean(), 2)
avg_calories   = round(df['calories'].mean(), 2)
avg_distance   = round(df['distance'].mean(), 4)
print('Avg Heart Rate:', avg_heart_rate)
print('Avg Steps:',      avg_steps)
print('Avg Calories:',   avg_calories)
print('Avg Distance:',   avg_distance)
```

```python
# Activity vs calories
act_cal = df.groupby('activity')['calories'].mean().reset_index()
print(act_cal)
```

```python
# Device vs steps
dev_steps = df.groupby('device')['steps'].mean().reset_index()
print(dev_steps)
```

```python
# Correlation heatmap
num_cols = ['age','height','weight','steps','hear_rate','calories','distance',
            'entropy_heart','entropy_setps','resting_heart','corr_heart_steps',
            'norm_heart','intensity_karvonen','sd_norm_heart','steps_times_distance']
plt.figure(figsize=(14,10))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Heart rate distribution by activity
plt.figure(figsize=(10,5))
for act in df['activity'].unique():
    subset = df[df['activity'] == act]['hear_rate']
    subset.plot(kind='kde', label=act)
plt.title('Heart Rate Distribution by Activity')
plt.legend()
plt.show()
```

```python
# Steps vs Calories scatter
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='steps', y='calories', hue='activity')
plt.title('Steps vs Calories')
plt.show()
```
